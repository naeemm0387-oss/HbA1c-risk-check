# 5. Merging NHANES Datasets
## 5.1 Load the prepared datasets

```python
import pandas as pd
# folder containing the prepared dataset

data_folder=r'E:\Bootcamp\99_Final-Projects\data\processed'

# load the CSV_files

demo= pd.read_csv(data_folder + r'\demo_prepared.csv')
bmx=pd.read_csv(data_folder + r'\bmx_prepared.csv')
diq= pd.read_csv(data_folder + r'\diq_prepared.csv')
ghb=pd.read_csv(data_folder + r'\ghb_prepared.csv')
```

```python
# Check dimensions
print("DEMO:", demo.shape)
print("BMX:", bmx.shape)
print("DIQ:", diq.shape)
print("GHB:", ghb.shape)
```

## 5.2 — Merge GHB with DEMO

```python
#  merging GHB with demographic data

merged_ghb_demo=ghb.merge(
    demo,
    on='SEQN',
    how='left'
)

# check the shape

print(merged_ghb_demo.shape)

#  checking missing age values

print(merged_ghb_demo['age_years'].isna().sum())

      # Display first five rows
display(merged_ghb_demo.head())

```

```python
merged_ghb_demo.duplicated().sum()
```

## 5.3 Merge with Body Measurements (BMX)

```python
merged_data= merged_ghb_demo.merge(
    bmx,
    on='SEQN',
    how='left'
)
# check the merge 

print(merged_data.shape)

# check missing bmi values

print(merged_data['bmi'].isna().sum())

# display the first five rows

display(merged_data.head())
```

## 5.4 Check Missing Values After Merging

```python
# check missing NA in merged data

merged_data.isna().sum()
```

## 5.5 — Check whether BMX participants matched

```python

# Check participants without a matching BMX record

unmatched_bmx = ~merged_data["SEQN"].isin(bmx["SEQN"])

# Count unmatched participants
print("Unmatched BMX participants:", unmatched_bmx.sum())
```

## 5.6 — Save the merged dataset

```python
# save merged datset

merged_data.to_csv(
    data_folder + r'\nhanes_merged.csv',
    index=False
)

#  check whether the file saved or not

from pathlib import Path

print(Path(data_folder + r'\nhanes_merged.csv').exists())

#  check the shape

print(merged_data.shape)
```
